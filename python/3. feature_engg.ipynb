# Feature Engineering

```python
df.columns
```

## Experience category (Fresher / Junior / Senior)

```python
df["years_of_experience"].value_counts()
```

```python
df["exp_category"] = df["years_of_experience"].apply(
    lambda x: 
    "Fresher" if x == 0 else 
    "Junior" if x in [1, 2] 
    else "Senior"
)
```

```python
df["exp_category"].value_counts()
```

---

## Academic performance bands
- 0-50   : Poor 
- 50-70  : Average  
- 70-85  : Good 
- 85-100 : Excellent    

```python
df["academic_score"] = round(df[['ssc_percentage', 'hsc_percentage', 'degree_percentage']].mean(axis=1), 3)
```

```python
df["academic_perfomance"] = df["academic_score"].apply(
    lambda x:
    "Poor" if 0 <= x < 50
    else "Average" if 50 <= x < 70
    else "Good" if 70 <= x <= 85
    else "Excellent"
)
```

---

## Skills match level (Low / Medium / High)

```python
df.head()
```

```python
df["skill_level"] = df["skills_match_percentage"].apply(
    lambda x: 
    "Low" if 0 <= x < 50
    else "Medium" if 50 <= x < 75
    else "High"
)
```

---

## Interview performance category

```python
df["interview_score"] = round(df[['technical_score', 'aptitude_score', 'communication_score']].mean(axis=1), 3)
```

```python
df["interview_performance"] = df["interview_score"].apply(
    lambda x:
    "Poor" if x < 50
    else "Average" if x < 65
    else "Good" if x < 80
    else "Excellent"
)
```
