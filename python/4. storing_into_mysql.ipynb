# Import Modules

```python
import pandas as pd
from sqlalchemy import create_engine, text
```

# Read CSV

```python
pd.set_option("display.max_columns", 30)
```

```python
df = pd.read_csv("../csv_files/hr_job_placement_dataset_cleaned.csv")
```

```python
df.shape
```

# Connecting to MYSQL - Using SQLAlchemy

```python
username = "root"
password = "root"
host = "localhost"
db_name = "job_acceptance_prediction"
```

```python
# Connecting to MySQL
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}"
)
```

```python
# Creating a new database 
conn = engine.connect()

try:
    conn.execute(text(
        f"CREATE DATABASE IF NOT EXISTS {db_name};"
    ))
    print(f"\"{db_name}\" Database Created Successfully.")
except:
    print("Error Occured While Creating Database.")
```

```python
# Reconnecting with database

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}/{db_name}"
)

conn = engine.connect()
```

# Create Table

```python
query = """
CREATE TABLE IF NOT EXISTS job_acceptance
(
    age_years                 INT,
    gender                    VARCHAR(10),
    ssc_percentage            DECIMAL(6, 3),
    hsc_percentage            DECIMAL(6, 3),
    degree_percentage         DECIMAL(6, 3),
    degree_specialization     VARCHAR(100),
    technical_score           DECIMAL(6, 3),
    aptitude_score            DECIMAL(6, 3),
    communication_score       DECIMAL(6, 3),
    skills_match_percentage   DECIMAL(6, 3),
    certifications_count      INT,
    internship_experience     VARCHAR(10),
    years_of_experience       INT,
    career_switch_willingness VARCHAR(20),
    relevant_experience       VARCHAR(20),
    previous_ctc_lpa          DECIMAL(6, 3),
    expected_ctc_lpa          DECIMAL(6, 3),
    company_tier              VARCHAR(20),
    job_role_match            VARCHAR(20),
    competition_level         VARCHAR(20),
    bond_requirement          VARCHAR(20),
    notice_period_days        INT,
    layoff_history            VARCHAR(20),
    employment_gap_months     INT,
    relocation_willingness    VARCHAR(20),
    status                    VARCHAR(20)
);
"""

try:
    conn.execute(text(query))
    conn.commit()

    print("Table Created Successfull!")
except:
    print("Error Occured while creating Table.")
```

# Inserting to MySQL

```python
df.to_sql(
    name="job_acceptance",
    con=engine,
    if_exists="append",
    index=False
)

conn.commit()
```

### Terminating MySQL Connection

```python
conn.close()
```
